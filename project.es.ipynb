#### Ejercicio 18. Normaliza la variable de precios para cada población y traza los 4 histogramas en el mismo gráfico (debes usar el subconjunto obtenido en la pregunta 14) (★★★)

Para el método de normalización, puedes usar el que consideres adecuado, no hay una única respuesta correcta para esta pregunta. Imprime el gráfico y escribe en la celda de Markdown un breve análisis sobre el gráfico.

Pista: Puedes ayudarte revisando la demostración multihist de Matplotlib.

```python
import pandas as pd
import matplotlib.pyplot as plt
ds = pd.read_csv('assets/real_estate.csv', sep=';')

valores_deseados = ["FuenLabrada", "Leganés", "Getafe", "Alcorcón"]
subset_ds = ds[ds['level5'].isin(valores_deseados)].copy()

# Verificando tamaños
print(subset_ds['level5'].value_counts())

# Normalizando precios por población (z-score)
subset_ds['price_norm'] = subset_ds.groupby('level5')['price'].transform(
    lambda x: (x - x.mean()) / x.std(ddof=1))

# Trazar histogramas
plt.figure(figsize=(10, 6))

# Para cada población, trazar un histograma con precios normalizados
for poblacion in valores_deseados:
    datos = subset_ds[subset_ds['level5'] == poblacion]['price_norm'].dropna()
    if datos.empty:
        continue
    plt.hist(datos, bins=20, alpha=0.5, label=f"{poblacion} (n={len(datos)})")

plt.title('Histograma de precios normalizados por población')
plt.xlabel('Precio normalizado (z-score)')
plt.ylabel('Frecuencia')
plt.legend(title='Población')
plt.grid(True)
plt.tight_layout()
plt.show()
```

Mientras que Getafe y Alcorcón tienen un alta de precios en la poblacion, Leganés es muy reducido en comparacion.

#### Ejercicio 20. Realiza el mismo gráfico para 4 poblaciones diferentes (columna level5) y colócalos en el mismo gráfico. Debes usar el subconjunto obtenido en la pregunta 14 (★★☆) 
Pista: Haz un diagrama de dispersión de cada población usando subgráficos (subplots).

```python
import pandas as pd
import matplotlib.pyplot as plt
ds = pd.read_csv('assets/real_estate.csv', sep=';')

poblaciones = ['Getafe', 'Alcorcón', 'Leganés', 'Fuenlabrada']

# Creando la figura y ejes del diagramaa: 2 filas x 2 columnas de subgráficos
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()  # sirve para aplanar e indexar más fácil

# Iterando sobre cada población y su subplot correspondiente
for i, poblacion in enumerate(poblaciones):
    data = subset_ds[subset_ds['level5'] == poblacion]
    
    # Scatter plot: superficie vs precio
    axes[i].scatter(data['surface'], data['price'], alpha=0.6, color='tab:blue', edgecolor='k')
    axes[i].set_title(f'{poblacion}')
    axes[i].set_xlabel('Superficie (m²)')
    axes[i].set_ylabel('Precio (€)')
    axes[i].grid(True)

# Ajustando espacio entre subplots
plt.tight_layout()
plt.suptitle('Relación entre Superficie y Precio en 4 Poblaciones', fontsize=16, y=1.02)
plt.show()
```
